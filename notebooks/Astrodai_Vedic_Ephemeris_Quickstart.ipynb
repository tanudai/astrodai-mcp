```python
# Install requests if running in a fresh Colab instance
!pip install -q requests pandas tabulate

import json
import urllib.request
import pandas as pd

ASTRODAI_MCP_URL = "https://astrodai.com/mcp"

def call_mcp(method: str, params: dict) -> dict:
    req = urllib.request.Request(
        ASTRODAI_MCP_URL,
        headers={"Content-Type": "application/json", "User-Agent": "Astrodai-Colab/1.0"},
        data=json.dumps({"jsonrpc": "2.0", "id": 1, "method": method, "params": params}).encode("utf-8")
    )
    with urllib.request.urlopen(req, timeout=15) as resp:
        return json.loads(resp.read().decode("utf-8"))

print("✅ Astrodai Client Initialized")
```

## 1. Resolve Location Coordinates via Astrodai Gazetteer Atlas

```python
city = "Bengaluru"
res = call_mcp("tools/call", {"name": "find_birthplace", "arguments": {"place": city}})
candidates = res["result"]["structuredContent"]["candidates"]
place_token = candidates[0]["place_token"]
label = candidates[0]["label"]

print(f"Resolved: {label}")
print(f"Place Token: {place_token[:30]}...")
```

## 2. Compute Arcsecond Sidereal Birth Chart (Kundli)

```python
birth_date = "1995-10-15"
birth_time = "14:30"

chart_res = call_mcp("tools/call", {
    "name": "chart_overview",
    "arguments": {"birth_date": birth_date, "birth_time": birth_time, "place_token": place_token}
})

print(chart_res["result"]["content"][0]["text"])
```

## 3. Career & Business Karma Analysis (Job vs Enterprise)

```python
career_res = call_mcp("tools/call", {
    "name": "life_area",
    "arguments": {"birth_date": birth_date, "birth_time": birth_time, "place_token": place_token, "area": "career"}
})

print(career_res["result"]["content"][0]["text"])
```
